```python
s="abcdefg hello"

print(s[::-1]) #step size is negative so go from backwards and print all

print(s[len(s)::-1]) #same as above ,keep on printing till end(including zero)

print(s[-1:-14:-1])

print(s[-1:-len(s)-1:-1])

print()

print(s[len(s):0:-1]) #prints till 0(zero excluded)

print(s[-1:0:-1]) #same as above

print(s[-1:-len(s):-1])

print()

print(s[1:len(s):1]) #starts from 1 and prints till len(s)(last index included)

print()

print(s[len(s)+1:0:-1])

print()

print(s[len(s):0:-2])

print()

print(s[len(s)+2:0:-1])

print(s[len(s)-2:0:-1])

print()

print(s[1:9:-2]) #this gives empty string because it doesnt make sense

print(s[9:1:-2])

print()

print(s[0::-1])

```
